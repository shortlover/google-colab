```python
import torch
import torch.nn as nn
import math

class GARCHNetLSTM(nn.Module):
    def __init__(self, input_size, hidden_size=64, num_lstm_layers=2, num_fc_layers=2):
        """
        GARCHNet model using LSTM layers

        Args:
            input_size: Number of input features
            hidden_size: Size of hidden layers
            num_lstm_layers: Number of LSTM layers
            num_fc_layers: Number of fully connected layers after LSTM
        """
        super().__init__()

        # LSTM layers for processing time series
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_lstm_layers,
            batch_first=True,
            dropout=0.1
        )

        # Fully connected layers
        self.fc_layers = nn.ModuleList()
        for i in range(num_fc_layers):
            if i == 0:
                self.fc_layers.append(nn.Linear(hidden_size, hidden_size))
            else:
                self.fc_layers.append(nn.Linear(hidden_size, hidden_size))

        # Output layers for variance, degrees of freedom, and skewness
        self.variance_layer = nn.Linear(hidden_size, 1)  # σt²
        self.dof_layer = nn.Linear(hidden_size, 1)      # η (nu)
        self.skewness_layer = nn.Linear(hidden_size, 1) # λ (lambda)

        self.dropout = nn.Dropout(0.1)

    def forward(self, x):
        """
        Forward pass

        Args:
            x: Input tensor of shape (batch_size, sequence_length, input_size)

        Returns:
            tuple: (variance, degrees_of_freedom, skewness)
        """
        # Process through LSTM layers
        lstm_out, _ = self.lstm(x)
        # Take the last output of the sequence
        last_out = lstm_out[:, -1, :]

        # Process through fully connected layers with dropout
        fc_out = last_out
        for fc_layer in self.fc_layers:
            fc_out = torch.relu(fc_layer(fc_out))
            fc_out = self.dropout(fc_out)

        # Compute outputs with appropriate activation functions
        # Variance (σt²) - using softplus for non-negativity
        variance = torch.log1p(torch.exp(self.variance_layer(fc_out)))

        # Degrees of freedom (η) - must be > 2
        dof = torch.log1p(torch.exp(self.dof_layer(fc_out))) + 2

        # Skewness (λ) - must be in (-1, 1)
        skewness = torch.tanh(self.skewness_layer(fc_out))

        return variance, dof, skewness

class HansenSkewedTNLLLoss(nn.Module):
    def __init__(self):
        super().__init__()

    def compute_ct(self, nu):
        """Compute ct coefficient for Hansen's skewed t-distribution"""
        return torch.exp(
            torch.lgamma((nu + 1) / 2) -
            torch.lgamma(nu / 2) -
            0.5 * torch.log(torch.pi * (nu - 2))
        )

    def forward(self, x, variance, dof, skewness):
        """
        Compute negative log-likelihood for Hansen's skewed t-distribution

        Args:
            x: Observed daily returns (Adj Close changes)
            variance: Predicted variance (σt²)
            dof: Degrees of freedom (η)
            skewness: Skewness parameter (λ)
        """
        # Compute standard deviation
        sigma = torch.sqrt(variance)

        # Compute standardized residuals (zero mean assumption)
        z = x / sigma

        # Compute ct coefficient
        ct = self.compute_ct(dof)

        # Compute log-likelihood
        squared_term = x**2 / ((dof - 2) * variance)
        sign_term = torch.sign(x)

        log_likelihood = torch.log(2) - 0.5 * torch.log(variance) + torch.log(ct) + (
            -(dof + 1) / 2 * torch.log1p(squared_term)
        ) + torch.log(torch.abs(sign_term))

        return -torch.mean(log_likelihood)

def train_garchnet(model, train_loader, val_loader, num_epochs, learning_rate=0.001):
    """
    Train the GARCHNet model

    Args:
        model: GARCHNet model instance
        train_loader: Training data loader
        val_loader: Validation data loader
        num_epochs: Number of training epochs
        learning_rate: Learning rate for optimization
    """
    criterion = HansenSkewedTNLLLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=5)

    best_val_loss = float('inf')

    for epoch in range(num_epochs):
        # Training phase
        model.train()
        train_loss = 0
        num_batches = 0

        for sequences, targets in train_loader:
            optimizer.zero_grad()

            # Get daily returns from targets (Adj Close changes)
            daily_returns = targets[:, targets.shape[1] - 1]  # Assuming Adj Close is the last feature

            # Forward pass
            variance, dof, skewness = model(sequences)

            # Compute loss
            loss = criterion(daily_returns, variance, dof, skewness)

            # Backward pass and optimize
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()

            train_loss += loss.item()
            num_batches += 1

        avg_train_loss = train_loss / num_batches

        # Validation phase
        model.eval()
        val_loss = 0
        num_val_batches = 0

        with torch.no_grad():
            for sequences, targets in val_loader:
                daily_returns = targets[:, targets.shape[1] - 1]
                variance, dof, skewness = model(sequences)
                loss = criterion(daily_returns, variance, dof, skewness)
                val_loss += loss.item()
                num_val_batches += 1

        avg_val_loss = val_loss / num_val_batches

        # Learning rate scheduling
        scheduler.step(avg_val_loss)

        # Save best model
        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            torch.save(model.state_dict(), 'best_garchnet_model.pth')

        print(f'Epoch [{epoch+1}/{num_epochs}]')
        print(f'Training Loss: {avg_train_loss:.6f}')
        print(f'Validation Loss: {avg_val_loss:.6f}')
        print(f'Learning Rate: {optimizer.param_groups[0]["lr"]:.6f}\n')

# Example usage
if __name__ == "__main__":
    # Assuming your dataset is already set up
    data_path = "/content/drive/MyDrive/Colab Notebooks/google-colab/data"
    window_size = 10
    batch_size = 32

    # Create datasets and dataloaders
    train_loader = create_dataloaders(data_path, window_size, batch_size, dataset_type='kdd17')
    val_loader = create_dataloaders(data_path, window_size, batch_size, dataset_type='kdd17')

    # Get number of features from a sample batch
    sample_sequences, _ = next(iter(train_loader))
    input_size = sample_sequences.shape[2]  # Number of features

    # Create and train model
    model = GARCHNetLSTM(input_size=input_size)
    train_garchnet(model, train_loader, val_loader, num_epochs=100)
```
